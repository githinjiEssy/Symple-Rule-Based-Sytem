```python
#a simple rule based chatbot

class RulebasedChtbot:
  def __init__(self):
    #Define rules for the chatbot
    self.rules = {
        "hello": "Hello! How can I help you today?",
        "hi": "Hi there! How can I assist you?",
        "bye": "Goodbye! Have a great day",
        "help": "I'm here to help. What do you need assistance with?"
    }
    self.default_response = "I'm sorry, I don't understand. Can you please rephrase your question?"

    def get_response(self, user_input):
      #check for the keywords n user input and return corresponding response
      for keyword, response in self.rules.items():
        if keyword in user_input.lower():
          return response
      return self.default_response

#example usage
if __name__ == "__main__":
  chatbot = RulebasedChtbot()

  while True:
    user_input = input("You: ")
    if user_input.lower() in ["exit", "quit"]:
      print("Chatbot: Goodbye! Have a good day")
      break
    response = chatbot.get_response(user_input)
    print("Chatbot:", response)
```
